```python
import zipfile
from pathlib import Path
import os
```

```python

out_dir = "data/explore"
file_path = "data/dataset.zip"

with zipfile.ZipFile(file_path, 'r') as zip_ref:
    for file in zip_ref.namelist():

        # Skip macOS metadata files
        if '__MACOSX' in file or os.path.basename(file).startswith('._'):
            continue

        dest_file = Path(out_dir)/file
        if dest_file.exists():
            print("skipping file already exists")
        else:
            print("extracting file")
            zip_ref.extract(file, out_dir)
```

```python

import pandas as pd

csv_file_path = "data/explore/assessments.csv"

try:
 
    df = pd.read_csv(csv_file_path)
    
    distinct_types = df["assessment_type"].dropna().unique()
    total_records = len(df)
    null_counts = df.isnull().sum()
    print(f"Total rows in file:        {total_records}")
    
    for column, count in null_counts.items():
        percentage = (count / total_records) * 100

        print(f"{column:<25} | {count:<12} ")
    df = df.head()
    print(f"First few rows of the dataset:\n{df}")

    print(f"Dataset shape: {df.shape}")
   

    df = df.describe()
    print(f"Descriptive statistics:\n{df}")
   

    df = df.isnull().sum()
    print(f"Missing values in each column:\n{df}")
   

    duplicates = df[df.duplicated()]
    print(f"Duplicate rows found: {duplicates}")
    
  
    print("distinct assessment types ")
    
    for assessment_type in distinct_types:
        print(f" - {assessment_type}")
        
except FileNotFoundError:
    print(f" Error: Could not find the file at {csv_file_path}")
except KeyError:
    print(" Error: The column 'assessment_type' does not exist in this CSV.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
